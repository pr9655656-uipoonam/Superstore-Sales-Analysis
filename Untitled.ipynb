```python
query = """
SELECT * FROM superstore
LIMIT 5;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity
FROM superstore;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Sales DESC;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Region
ORDER BY Total_Sales DESC;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    "Sub-Category",
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY "Sub-Category"
ORDER BY Total_Profit DESC;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    year,
    month_name,
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY year,month_name
ORDER BY year;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    "Product Name",
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 5;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    "Product Name",
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY "Product Name"
ORDER BY Total_Profit DESC
LIMIT 5;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    "Product Name",
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY "Product Name"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    Segment,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Segment
ORDER BY Total_Sales DESC;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT 
    Discount,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Discount
ORDER BY Discount;
"""
result = pd.read_sql_query(query,conn)
result
```

# outliers detection

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df[['Sales','Profit','Quantity','Discount']])
plt.savefig(
    "boxplot.png",
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
num_cols = ['Sales','Profit','Quantity','Discount']

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df[['Sales','Profit','Quantity','Discount']])
plt.savefig(
    "boxesplot.png",
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

# Forecasting

```python
monthly_sales = df.groupby(pd.Grouper(key='Order Date',freq='M'))['Sales'].sum()
```

```python
train = monthly_sales[:-6]
test = monthly_sales[-6:]
```

```python
forecast = [train.mean()]*len(test)
```

# Model Evaluation

```python
# MAE
mae = mean_absolute_error(test,forecast)
print(mae)
```

```python
#RMSE
rmse = np.sqrt(mean_squared_error(test,forecast))
print(rmse)
```
